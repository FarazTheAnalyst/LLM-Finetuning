```python
from trl import DPOTrainer
from transfromer import AutoTokenizer, AutoModelForCausalLM, TrainingArguments
from peft import PeftModel
from peft import LoraConfig, get_peft_model, TaskType
from datasts import load_dataset
improt torch
```

```python
base_model = "TinyLlama/TinyLlama-1.1B-intermediate-step-1431k-3T"
```

```python
instruction_checkpoint = "/content/checkpoint-3"
```

```python
# Load dataset
dataset = load_dataset("csv", data_files="/content/pharma_preference_data.csv")["train"]
```

```python
tokenizer = AutoTokenizer.from_pretrained(base_model)
```

```python
if tokenizer.pad_token is None:
  tokenizer.pad_token = tokenizer.eos_token
```

get_peft_model() -> Create a New LoRA during training

PeftModel.from_pretrained() -> Load an already-trained LoRA for inference or fruther training

```python
lora_config = LoraConfig(
    task_type=TaskType.CAUSAL_LM,
    r=8,
    lora_alpha=16,
    lora_dropout=0.05,
    target_modules=["q_proj", "v_proj"],
    bias="none"
)
```

```python
# pref_model_lora = get_peft_model(instruction_model, lora_config)
```

```python
base_model
```

```python
# STEP A: Load base
model = AutoModelForCausalLM.from_pretrained(
    base_model,
    load_in_8bit=True,
    device_map="auto"
)
```

```python
# STEP B: Load Instruction LoRA + merge
model = PeftModel.from_pretrained(model, instruction_checkpoint)
```

```python
model = model.merge_and_unload()
```

```python
# STEP C: Attach NEW LoRA for preference
pref_model_lora = get_peft_model(model, lora_config)
```

```python
import os
os.environ["WANDB_DISABLED"] = "true"
```

```python
from trl import DPOTrainer, DPOConfig
```

```python
dpo_args = DPOConfig(
    output_dir="./tinyllama-preference-alignment",
    learning_rate=2e-5,
    per_device_train_batch_size=1,
    gradient_accumulation_steps=8,
    num_train_epochs=1,
    beta=0.1,
    report_to=None,
    logging_dir=None, # disable logging to wandb or tensorboard
    loss_type="sigmoid", # or "hinge", depending on expeirment
    remove_unused_columns=False
)
```

```python
trainer = DPOTrainer(
    model=pref_model_lora,
    ref_model="None",
    args=dpo_args,
    train_datset=dataset,
    processing_class=tokenizer, # instead of tokenizer argument
    # you can pass data_collator if needed,
    # optionally eval_datset etc.
)
```

```python
trainer.train()
```
